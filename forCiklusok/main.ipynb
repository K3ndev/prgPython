```python
import time
from rich.console import Console
from rich.table import Table
from random import randint

def safeInput(text, t=int, sign=0, minimum=None, maximum=None):
    while True:
        try:
            value = t(input(text))
        except ValueError:
            print(f"Kérem {t}-t adjon meg!")
            continue

        if sign == 1 and value <= 0:
            print("kérem adjon meg pozitív számot")
        elif sign == -1 and value >= 0:
            print("kérem adjon meg negatív számot")
        elif minimum is not None and value < minimum:
            print(f"Az érték legalább {minimum} legyen")
        elif maximum is not None and value > maximum:
            print(f"Az érték legfeljebb {maximum} legyen")
        else:
            return value
```

```python
print("------- 15. Feladat -------")

def isPrime(num):
    if num < 2:
        return False
    divisor = 2
    while divisor * divisor <= num:
        if num % divisor == 0:
            return False
        divisor += 1
    return True

n = safeInput("->", minimum=0)
j = 2
for _ in range(n):
    while not isPrime(j):
        j += 1
    print(j)
    j += 1
```

```python
print("------- 23. Feladat -------")

data = {"g": 0, "b": 0}
for i in range(10):
    a = randint(1, 100)
    b = randint(1, 100)
    ösz = safeInput(f"{a} + {b} = ")
    kül = safeInput(f"{a} - {b} = ")
    isGood = True
    if ösz == a + b:
        print(f"{a} + {b} = {ösz}")
        print("ez sikerült")
    else:
        print(f"{a} + {b} != {ösz}")
        print("ez nem sikerült")
        isGood = False
    if kül == a - b:
        print(f"{a} - {b} = {kül}")
        print("ez sikerült")
    else:
        print(f"{a} - {b} != {kül}")
        print("ez nem sikerült")
        isGood = False
    if isGood:
        print("Ez a feladatsorozat sikerült")
        data["g"] += 1
    else:
        print("Ez a feladatsorozat nem sikerült")
        data["b"] += 1
print(f"10 feladatból {data['b']} rontottál el és {data['g']}")
```
